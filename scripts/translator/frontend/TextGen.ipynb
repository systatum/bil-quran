```python
import json
import re
from typing import Iterable
```

```python
def is_exegesis_empty(text: str) -> bool:
    return re.match(r'^See <\{\[.*?\]\}>$', text) is not None
```

```python
def get_quran_ayahs() -> Iterable[int]:
    return range(1, 115)
```

```python
texts = []
for i in range(1, 115):
    data = json.load(open("../../../public/quran/exegesis/mirali/en-US/{}.json".format(i)))
    texts.extend((f"Exegesis mirali surah {i} ayah {k}", v) for k, v in data["exegesis"].items() if not is_exegesis_empty(v))
    texts.extend((f"Translation mirali surah {i} ayah {k}", v) for k, v in data["translations"].items())

    data = json.load(open("../../../public/quran/exegesis/ibnkathir/en-US/{}.json".format(i)))
    texts.extend((f"Exegesis ibnkathir surah {i} ayah {k}", v) for k, v in data["exegesis"].items() if not is_exegesis_empty(v))
    texts.extend((f"Translation ibnkathir surah {i} ayah {k}", v) for k, v in data["translations"].items())

    data = json.load(open("../../../public/quran/exegesis/aliquli/en-US/{}.json".format(i)))
    texts.extend((f"Translation aliquli surah {i} ayah {k}", v) for k, v in data["translations"].items())
len(texts)
```
